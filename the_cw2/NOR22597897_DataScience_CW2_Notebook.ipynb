## 1. Load and Inspect Dataset

In this section, the dataset is loaded using pandas and basic inspection is performed. 
This includes viewing the first few rows, checking data types, and analysing basic statistics 
to understand the structure and quality of the dataset before further processing.

```python
# Import required libraries
import pandas as pd

# Import function to split dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Import scaler to normalize feature values between 0 and 1
from sklearn.preprocessing import MinMaxScaler

# Import Naive Bayes model
from sklearn.naive_bayes import GaussianNB

# Import evaluation metrics
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Import visualization libraries
import seaborn as sns
import matplotlib.pyplot as plt

# Import Logistic Regression model
from sklearn.linear_model import LogisticRegression


# Load dataset (cleaned version with updated Pass threshold)
df = pd.read_excel("NOR22597897_CleanedDataset_v2.xlsx")

# Display first few rows to understand the dataset
df.head()

# Check structure, column names, and data types
df.info()

# Display basic statistics for numerical columns
df.describe()
```

## 2. Explore Target Variable (Exam Score)

Before creating the classification target, the distribution of the Exam_Score variable is analysed. 
This helps to understand how scores are distributed and whether the initial threshold for passing is appropriate.

```python
# Analyse distribution of exam scores

# Summary statistics
df["Exam_Score"].describe()

# Histogram to visualize distribution
df["Exam_Score"].hist()
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Frequency")
plt.show()
```

## 3. Preprocessing: Pass Variable Adjustment

The Pass variable is redefined using a threshold of 70. 
The original threshold (50) resulted in all students being classified as pass, 
which made the dataset unsuitable for classification.

This adjustment creates a more balanced dataset and allows the model to learn meaningful patterns.

```python
# Redefine Pass variable using a threshold of 70
df["Pass"] = (df["Exam_Score"] >= 70).astype(int)

# Save updated dataset (optional - for consistency)
# df.to_excel("NOR22597897_CleanedDataset_v2.xlsx", index=False)

# Check distribution of Pass variable
df["Pass"].value_counts()
```

## 4. Feature Selection

In this step, the dataset is split into input features (X) and the target variable (y).  
The Exam_Score column is removed from the features to prevent data leakage, as it directly determines the Pass variable.

```python
# Separate features and target variable

# X contains all input features except Pass and Exam_Score
X = df.drop(columns=["Pass", "Exam_Score"])

# y contains the target variable (Pass)
y = df["Pass"]
```

## 5. Train-Test Split

The dataset is split into training and testing sets using an 80/20 ratio.  
The training set is used to train the model, while the testing set is used to evaluate its performance on unseen data.

```python
# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,   # 20% test data
    random_state=42  # ensures reproducibility
)
```

## 6. Encoding Categorical Variables

Categorical features are converted into numerical format using one-hot encoding. 
This is necessary because machine learning models cannot process text-based data directly.

After encoding, the training and testing datasets are aligned to ensure they have the same columns.

```python
# Convert categorical features into numerical format using one-hot encoding
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

# Ensure both training and testing sets have the same columns
# Any missing columns in X_test are filled with 0
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)
```

## 7. Feature Scaling

Feature scaling is applied using Min-Max normalisation. 
This transforms all values into a range between 0 and 1, ensuring that features with larger magnitudes 
do not dominate the model.

The scaler is fitted only on the training data and then applied to the test data to avoid data leakage.

```python
# Create scaler object
scaler = MinMaxScaler()

# Fit the scaler on training data and transform it
X_train = scaler.fit_transform(X_train)

# Apply the same transformation to test data (without fitting again)
X_test = scaler.transform(X_test)
```

## 8. Naive Bayes Model Training

The Naive Bayes model is trained using the training dataset. 
This model was selected in Coursework 1 due to its simplicity and efficiency for classification tasks.

```python
# Create Naive Bayes model instance
model = GaussianNB()

# Train the model using training data
model.fit(X_train, y_train)
```

## 9. Model Prediction

The trained Naive Bayes model is used to predict outcomes on the test dataset.

```python
# Predict outcomes on test data
y_pred = model.predict(X_test)
```

## 10. Model Evaluation

The model performance is evaluated using accuracy, confusion matrix, and classification report. 
These metrics provide insight into how well the model is performing.

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# Confusion matrix
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

# Classification report (precision, recall, F1-score)
print("Classification Report:\n", classification_report(y_test, y_pred))
```

## 11. Confusion Matrix Visualization

The confusion matrix is visualized using a heatmap to provide a clearer understanding of model performance.

```python
# Create confusion matrix heatmap
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Naive Bayes Confusion Matrix")
plt.show()
```

## 12. Logistic Regression Model

Logistic Regression is implemented as an additional model to further analyse the dataset. 
This model helps to understand the relationship between features and the target variable.

```python
# Create Logistic Regression model instance
log_model = LogisticRegression(max_iter=1000)

# Train the model on training data
log_model.fit(X_train, y_train)
```

## 13. Logistic Regression Prediction

The trained Logistic Regression model is used to make predictions on the test dataset.

```python
# Predict outcomes on test data
y_pred_log = log_model.predict(X_test)
```

## 14. Logistic Regression Evaluation

The performance of the Logistic Regression model is evaluated using accuracy, confusion matrix, and classification report.

```python
# Calculate accuracy
log_accuracy = accuracy_score(y_test, y_pred_log)
print("Logistic Regression Accuracy:", log_accuracy)

# Confusion matrix
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_log))

# Classification report
print("Classification Report:\n", classification_report(y_test, y_pred_log))
```

## 15. Feature Importance Analysis

Logistic Regression provides insight into how each feature influences the prediction through its coefficients. 
Higher coefficient values indicate a stronger impact on the likelihood of a student passing.

```python
# Get feature importance (coefficients)
feature_names = X.columns
coefficients = log_model.coef_[0]

# Create DataFrame for better visualization
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients[:len(feature_names)]
})

# Sort features by importance
feature_importance = feature_importance.sort_values(by="Coefficient", ascending=False)

# Display top 10 most important features
feature_importance.head(10)
```

## 16. Logistic Regression Confusion Matrix

The confusion matrix is visualized to clearly show how well the Logistic Regression model performs.

```python
# Create confusion matrix heatmap for Logistic Regression
cm_log = confusion_matrix(y_test, y_pred_log)

sns.heatmap(cm_log, annot=True, fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression Confusion Matrix")
plt.show()
```
